## 1. Mathematical Background

### Leaky Integrate-and-Fire (LIF) Dynamics
The membrane potential $U[t]$ of a discrete-time Leaky Integrate-and-Fire neuron evolves as:
$$U[t] = \beta U[t-1] + W X[t] - S[t-1] U_{\text{thr}}$$
where $\beta \in (0, 1)$ is the membrane decay rate, $W$ represents synaptic weights, $X[t]$ is the input, and $S[t]$ is the output spike:
$$S[t] = \Theta(U[t] - U_{\text{thr}}) = \begin{cases} 1, & U[t] \ge U_{\text{thr}} \\ 0, & U[t] < U_{\text{thr}} \end{cases}$$

### Surrogate Gradient Backpropagation
Because the non-differentiable Heaviside step function $\Theta(\cdot)$ has zero derivative everywhere except at $0$ where it is undefined ($\rac{\partial S}{\partial U} = \delta(U)$), we substitute a smooth surrogate derivative during the backward pass:

**Gaussian Surrogate Gradient:**
$$\frac{\partial S}{\partial U} = \frac{\sigma}{\sqrt{2\pi}} \exp\left( -\frac{\sigma^2 U^2}{2} \right)$$

**Smoothed Heaviside Surrogate Gradient:**
$$\frac{\partial S}{\partial U} = \frac{\alpha}{2 \left( 1 + \alpha |U| \right)^2}$$

### Bio-Plausible KL-Divergence Firing Rate Penalty
To enforce energy-efficient sparse spiking (similar to biological neocortex), we penalize the Kullback-Leibler divergence between the actual average firing rate $\hat{\rho}$ and target rate $\rho_{\text{target}}$:
$$\mathcal{L}_{\text{KL}} = \lambda \sum_{i} \left[ \rho_{\text{target}} \log \frac{\rho_{\text{target}}}{\hat{\rho}_i + \epsilon} + (1 - \rho_{\text{target}}) \log \frac{1 - \rho_{\text{target}}}{1 - \hat{\rho}_i + \epsilon} \right]$$

```python
import torch
import torch.nn as nn
    import snntorch as snn
from snntorch import surrogate
from snntorch import functional as SF

print(f"PyTorch Version: {torch.__version__}")
print(f"snnTorch Version: {snn.__version__}")
```

## 2. Defining an SNN Architecture with Custom Surrogates

```python
class BioSNN(nn.Module):
    def __init__(self, num_inputs=28, num_hidden=128, num_outputs=10, beta=0.9, spike_grad=None):
        super(BioSNN, self).__init__()
        if spike_grad is None:
            spike_grad = surrogate.gaussian(sigma=1.5)
            
        self.fc1 = nn.Linear(num_inputs, num_hidden)
        self.lif1 = snn.Leaky(beta=beta, spike_grad=spike_grad, init_hidden=True)
        self.fc2 = nn.Linear(num_hidden, num_outputs)
        self.lif2 = snn.Leaky(beta=beta, spike_grad=spike_grad, init_hidden=True, output=True)
        
    def forward(self, x_seq):
        # x_seq shape: [time_steps, batch_size, num_inputs]
        spk1_rec = []
        spk2_rec = []
        mem2_rec = []
        
        # Reset neuron states
        snn.utils.reset(self)
        
        for step in range(x_seq.size(0)):
            cur1 = self.fc1(x_seq[step])
            spk1 = self.lif1(cur1)
            cur2 = self.fc2(spk1)
            spk2, mem2 = self.lif2(cur2)
            
            spk1_rec.append(spk1)
            spk2_rec.append(spk2)
            mem2_rec.append(mem2)
            
        return torch.stack(spk1_rec), torch.stack(spk2_rec), torch.stack(mem2_rec)
```

## 3. Training with KL Firing Rate Sparsity & Temporal Regularization

```python
# Hyperparameters
num_steps = 25
batch_size = 32
model = BioSNN(spike_grad=surrogate.smoothed_heaviside(alpha=2.0))
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# Loss Functions
ce_loss = SF.ce_rate_loss()
kl_reg = SF.kl_rate_sparsity(target_rate=0.05, Lambda=1e-3)
temp_reg = SF.temporal_sparsity(Lambda=1e-4)

# Synthetic Spike Sequence Data
dummy_inputs = torch.rand(num_steps, batch_size, 28)
dummy_targets = torch.randint(0, 10, (batch_size,))

print("Beginning SNN Training Loop...")
for epoch in range(5):
    optimizer.zero_grad()
    spk1_rec, spk2_rec, mem2_rec = model(dummy_inputs)
    
    # Compute Loss Components
    task_loss = ce_loss(spk2_rec, dummy_targets)
    sparsity_loss = kl_reg(spk1_rec)
    burst_loss = temp_reg(spk1_rec)
    
    total_loss = task_loss + sparsity_loss + burst_loss
    total_loss.backward()
    optimizer.step()
    
    avg_firing_rate = spk1_rec.mean().item()
    print(f"Epoch {epoch+1:02d} | Task Loss: {task_loss.item():.4f} | KL Sparsity: {sparsity_loss.item():.4f} | Avg Fire Rate: {avg_firing_rate*100:.2f}%")
```
